```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('/content/student_exam_performance.csv')

# Display the first five rows
df.head()
```

```python

# Number of rows and columns
print("Dataset shape:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Dataset information
print("\nDataset information:")
df.info()

# Missing values
print("\nMissing values:")
print(df.isnull().sum())
```

```python

# Check dataset size
print("Number of rows and columns:", df.shape)

# Display all column names
print("\nColumn names:")
print(df.columns.tolist())

# Check data types
print("\nData types:")
print(df.dtypes)

# Check missing values
print("\nMissing values:")
print(df.isnull().sum())

# Summary statistics
print("\nNumerical summary:")
display(df.describe())
```

```python

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Final_Exam_Score",
    bins=10,
    kde=True
)

plt.title("Distribution of Final Exam Scores")
plt.xlabel("Final Exam Score")
plt.ylabel("Number of Students")
plt.show()
```

```python

# Count students in each Pass/Fail category
print(df["Pass_Fail"].value_counts())

# Create a bar chart
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="Pass_Fail")

plt.title("Distribution of Pass and Fail Students")
plt.xlabel("Pass/Fail Status")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

# Calculate percentages of each category
pass_fail_percent = df["Pass_Fail"].value_counts(normalize=True) * 100

print("Pass/Fail percentages:")
print(pass_fail_percent.round(2))
```

```python

# Select numerical columns for boxplots
num_cols = [
    "Study_Hours_Per_Week",
    "Attendance_Percentage",
    "Previous_Score",
    "Sleep_Hours",
    "Final_Exam_Score"
]

# Draw boxplots
plt.figure(figsize=(12, 6))

for i, col in enumerate(num_cols, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=df[col])
    plt.title(col.replace("_", " "))

plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Study_Hours_Per_Week",
    y="Final_Exam_Score",
    hue="Pass_Fail",
    s=100
)

plt.title("Study Hours vs Final Exam Score")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Final Exam Score")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Attendance_Percentage",
    y="Final_Exam_Score",
    hue="Pass_Fail",
    s=100
)

plt.title("Attendance vs Final Exam Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Final Exam Score")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Previous_Score",
    y="Final_Exam_Score",
    hue="Pass_Fail",
    s=100
)

plt.title("Previous Score vs Final Exam Score")
plt.xlabel("Previous Score")
plt.ylabel("Final Exam Score")
plt.tight_layout()
plt.show()
```

```python

# Select numerical columns
numeric_df = df.select_dtypes(include="number").drop(
    columns=["Student_ID"],
    errors="ignore"
)

# Calculate correlation
correlation_matrix = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Study_Hours_Per_Week",
    y="Attendance_Percentage",
    hue="Final_Exam_Score",
    size="Final_Exam_Score",
    palette="viridis",
    sizes=(40, 200)
)

plt.title("Multivariate Analysis of Student Performance")
plt.xlabel("Study Hours Per Week")
plt.ylabel("Attendance Percentage")
plt.tight_layout()
plt.show()
```

```python

# Check the lowest and highest final exam scores
print("Lowest final exam scores:")
display(df.nsmallest(5, "Final_Exam_Score"))

print("Highest final exam scores:")
display(df.nlargest(5, "Final_Exam_Score"))

# Check possible outliers using the IQR method
col = "Final_Exam_Score"

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df[col] < lower_bound) |
    (df[col] > upper_bound)
]

print("Potential outliers:", len(outliers))
display(outliers)
```
